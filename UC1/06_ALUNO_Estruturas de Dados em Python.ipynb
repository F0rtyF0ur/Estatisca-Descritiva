## 1. LISTAS
Conceito: Lista é uma coleção ordenada e mutável.
  
Características:  
✅ Ordenada  
✅ Permite alterações  
✅ Permite valores repetidos  
✅ Muito utilizada em Data Science  

### Índices:  
Todo elemento possui uma posição.



```python
frutas = [
    "Maçã",
    "Banana",
    "Laranja"
]
```


| Índice | Valor   |
| ------ | ------- |
| 0      | Maçã    |
| 1      | Banana  |
| 2      | Laranja |


### Acessando Elementos

```python
print(frutas[0])
print(frutas[1])
print(frutas[2])
```

#### Índices Negativos:

```python
print(frutas[-1])
print(frutas[-2])
print(frutas[-3])   
```

#### Método append():  
Adiciona um elemento ao final.

```python
frutas = ["Maçã","Banana"]
frutas.append("Laranja")

print(frutas)
```

#### Método insert()
Adiciona em posição específica.

```python
frutas = ["Maçã","Banana"]

frutas.insert(1,"Uva")

print(frutas)
```

#### Método remove()   
Remove pelo valor.

```python
frutas.remove("Banana")
```

#### Método pop()  
Remove pela posição.

```python
frutas.pop(0)
```

#### Método len()
Retorna quantidade de elementos.

```python
frutas = [
    "Maçã",
    "Banana",
    "Laranja"
]

print(f"O número de frutas na lista é: {len(frutas)}")
```

#### Método sort()
Ordena a lista.

```python
nomes = [
    "Carlos",
    "Ana",
    "Pedro"
]

nomes.sort()

print(f"Os nomes em ordem alfabética são: {nomes}")
```

#### Percorrendo Lista com For

```python
frutas = [
    "Maçã",
    "Banana",
    "Laranja"
]

for fruta in frutas:
    print(fruta)
```

#### Exemplo utilização de lista com input

```python
# 1. Inicializar uma lista vazia para armazenar os nomes
lista_nomes = []

print("--- Cadastro de Nomes ---")
print("Por favor, digite os nomes. Digite 'fim' para terminar o cadastro.")

# 2. Loop para coletar os dados do usuário continuamente
while True:
    # Pede a entrada do usuário
    nome_digitado = input("Digite um nome: ")

    # Verifica se o usuário digitou 'fim' para sair do loop
    if nome_digitado.lower() == 'fim':
        break

    # 3. Adiciona o nome coletado à lista
    lista_nomes.append(nome_digitado)

# 4. Exibir o resultado final
print("\n=====================================")
print("Lista de Nomes Cadastrados:")
print(lista_nomes)
print("=====================================")
```

União de conjuntos.


```python
turma_a = {"Ana","Pedro"}

turma_b = {"Carlos","João"}

resultado = turma_a.union(turma_b)

print(resultado)
```

### Dicionário aninhado

```python
alunos = {
    "aluno_1": {
        "nome": "Maria Silva",
        "idade": 20,
        "serie": "Ensino Médio",
        "nota_media": 8.5
    },
    "aluno_2": {
        "nome": "João Santos",
        "idade": 18,
        "serie": "Ensino Médio",
        "nota_media": 7.9
    }
}

print(alunos)
```

```python
for aluno_id, dados in alunos.items():
    print(f"ID do aluno: {aluno_id}") # Imprime o ID do aluno
    print(f"Nome: {dados['nome']}, Nota: {dados['nota_media']}") # Imprime o nome e a nota do aluno
```

#### Dicionário Com INPUT

```python
# 1. Inicializar um dicionário vazio para armazenar os dados
cadastro_pessoas = {}

print("--- Cadastro de Pessoas (Dicionário) ---")
print("Por favor, digite o nome e a idade. Digite 'fim' para terminar o cadastro.")

# 2. Loop para coletar os dados do usuário continuamente
while True:
    # Coletar o nome
    nome = input("\nDigite o NOME da pessoa (digite 'fim' para sair): ")
    
    if nome.lower() == 'fim':
        break

    # Coletar a idade (o programa tentará converter diretamente, podendo falhar se o usuário digitar texto)
    idade_input = int(input("Digite a IDADE da pessoa: "))
    
    idade = idade_input # Aqui, você pode adicionar validação para garantir que a idade seja um número inteiro válido, se desejar.
    cadastro_pessoas[nome] = idade # Adiciona o nome e a idade ao dicionário
    print(f"✅ {nome} cadastrado com sucesso!")    
    

# 3. Exibir o resultado final
print("\n=====================================")
print("Cadastro Final de Pessoas:")

if cadastro_pessoas:
    for nome, idade in cadastro_pessoas.items():
        print(f"Nome: {nome}, Idade: {idade}")
else:
    print("Nenhum dado foi cadastrado.")
    
print("=====================================")
```

Estruturas de Dados Fundamentais em Python

| Estrutura | Característica Principal | Mutabilidade | Ordem dos Elementos | Permite Duplicatas? | 
| :--- | :--- | :--- | :--- | :--- | 
| Listas (list) | Coleção ordenada de itens. | Mutável (Pode ser alterada) | Sim | Sim | 
| Tuplas (tuple) | Coleção ordenada de itens. | Imutável (Não pode ser alterada) | Sim | Sim | 
| Conjuntos (set) | Coleção não ordenada de itens únicos. | Mutável (Pode adicionar/remover elementos) | Não (em versões antigas), mas a ordem de inserção é mantida em versões recentes. | Não (Elementos devem ser únicos) |
| Dicionários (dict) | Coleção de pares chave-valor (key-value). | Mutável (Pode adicionar/remover pares) | Sim (a partir do Python 3.7+) | As chaves (keys) devem ser únicas e imutáveis. |

Detalhamento das Funções:
1. Listas (list)  
Função: Usadas para armazenar coleções de itens que precisam ser modificados frequentemente.  
Exemplo: [1, "a", 3.0]  
Uso Comum: Adicionar, remover, modificar elementos.  
2. Tuplas (tuple)  
Função: Usadas para armazenar coleções de itens que devem permanecer fixos e não serem alterados (dados imutáveis).  
Exemplo: (1, "a", 3.0)  
Uso Comum: Coordenadas, valores de retorno de funções onde a integridade dos dados é importante.  
3. Conjuntos (set)  
Função: Usadas para armazenar coleções de itens únicos. São ideais para operações matemáticas de conjuntos (união, interseção).  
Exemplo: {1, "a", 3.0}  
Uso Comum: Verificar a presença de um item rapidamente, remover duplicatas de uma lista.  
4. Dicionários (dict)  
Função: Usadas para armazenar dados em formato de pares chave-valor. A chave é usada para acessar o valor correspondente.  
Exemplo: {"nome": "Maria", "idade": 20}  
Uso Comum: Representar objetos com atributos, configurações ou registros onde a identificação (chave) é mais importante que a ordem.
