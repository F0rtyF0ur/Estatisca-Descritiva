## Manipulação de Arquivos CSV com Pandas em Python

O [Pandas](https://pandas.pydata.org/docs/ "Pandas Documentation") é a biblioteca mais popular do Python para análise e manipulação de dados. Este tutorial vai te guiar desde os conceitos mais básicos até técnicas intermediárias para trabalhar com arquivos **CSV**.

---

## 1. O Básico: Leitura e Escrita

Para começar, precisamos importar a biblioteca e entender como abrir e salvar arquivos.

### 1.1 Como ler um arquivo CSV
A função principal para carregar dados é a `pd.read_csv()`.

```python
import pandas as pd

# Lendo um arquivo CSV básico
df = pd.read_csv("dados.csv")

# Exibindo as primeiras 5 linhas do DataFrame
print(df.head())
```

### 1.2 Como salvar dados em um arquivo CSV
Para exportar os dados manipulados de volta para um arquivo CSV, utilizamos o método `.to_csv()`.

```python
# Criando um DataFrame de exemplo
dados = {
    'Nome': ['Ana', 'Bruno', 'Carlos'],
    'Idade': [23, 35, 41],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte']
}
df_exemplo = pd.DataFrame(dados)

# Salvando em um arquivo CSV (index=False evita criar uma coluna extra com os índices)
df_exemplo.to_csv("novos_dados.csv", index=False)
```

---

````python
# Criando um DataFrame de exemplo
import pandas as pd

try:
    
    dados = {
        'Nome': ['Ana', 'Bruno', 'Carlos'],
        'Idade': [23, 35, 41],
        'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte']
    }
    df_exemplo = pd.DataFrame(dados) # Criando o DataFrame a partir do dicionário de dados

    # Salvando em um arquivo CSV (index=False evita criar uma coluna extra com os índices)
    df_exemplo.to_csv("novos_dados.csv", index=False) # Salvando o DataFrame em um arquivo CSV chamado "novos_dados.csv"
    print("Exportação do arquivo .csv realizada com sucesso!")
except Exception as e:
    print(f"Erro ao exportar o arquivo .csv: {e}")
````

### 2. Lidando com Arquivos CSV Problemáticos

Nem todo arquivo CSV é perfeito. Muitas vezes eles usam delimitadores diferentes ou codificações específicas.

#### 2.1 Alterando o Delimitador e a Codificação
No Brasil, é muito comum que arquivos CSV usem ponto e vírgula (`;`) como separador (já que usamos a vírgula para decimais) e codificação `utf-8` ou `latin1`.

```python
# Lendo um CSV separado por ponto e vírgula e com codificação específica
df = pd.read_csv("dados_brasil.csv", sep=";", encoding="utf-8")
```

#### 2.2 Selecionando colunas específicas na leitura
Se o arquivo for muito grande e você só precisar de algumas colunas, use o argumento `usecols`.

```python
# Carrega apenas as colunas 'Nome' e 'Cidade'
df = pd.read_csv("dados.csv", usecols=['Nome', 'Cidade'])
```


Exemplo:

````python
import pandas as pd

# 1. Tentativa errada (comum): Se você ler direto, o Pandas não entenderá as colunas por causa do ';'
# df_errado = pd.read_csv('generated/dados_funcionarios.csv') 

# 2. Forma Correta: Ajustando o separador e especificando colunas de interesse
# Vamos carregar apenas as colunas: 'Nome', 'Cidade' e 'Departamento'
colunas_filtradas = ['Nome', 'Cidade', 'Departamento']

df = pd.read_csv(
    'Arquivos_CSV/dados_funcionarios.csv', 
    sep=';',                     # Define o ponto e vírgula como separador das colunas
    encoding='utf-8',            # Garante que caracteres como 'São Paulo' e 'João' sejam lidos sem erros
    usecols=colunas_filtradas    # Otimiza a memória carregando apenas o necessário
)

# Exibindo o resultado na tela
print("--- Dados importados com sucesso ---")
print(df)
````

#### 3.2 Filtrando linhas com base em condições
Assim como você forçou condições nos erros anteriormente, você pode criar filtros condicionais nos dados:

```python
# Filtrando pessoas com mais de 30 anos
maiores_de_30 = df[df['Idade'] > 30]

# Múltiplas condições: Maiores de 30 anos E que moram em 'São Paulo'
# Use & para 'E' / Use | para 'OU' (e sempre coloque as condições entre parênteses)
filtro_composto = df[(df['Idade'] > 30) & (df['Cidade'] == 'São Paulo')]
```


Exemplo:



Filtrando Dados  

No Pandas, você filtra linhas criando uma condição lógica dentro de colchetes df[condição]. Veja como selecionar apenas os funcionários do departamento de TI:

````python
import pandas as pd

# Carrega o arquivo completo
df = pd.read_csv('funcionarios.csv', sep=',', encoding='utf-8')

# Criando o filtro para o departamento de TI
filtro_ti = df['Cargo'] == 'Assistente'
df_ti = df[filtro_ti]

print("--- Funcionários do Departamento de TI ---")
print(df_ti)
````

### 4. Tratamento de Dados Ausentes (Valores Nulos)

Arquivos CSV do mundo real frequentemente possuem dados faltando (representados como `NaN` no Pandas).

```python
# Verificar quantos valores nulos existem por coluna
print(df.isnull().sum())

# Opção 1: Remover todas as linhas que possuem qualquer valor nulo
df_limpo = df.dropna()

# Opção 2: Substituir os valores nulos por um valor padrão (ex: 0 ou 'Desconhecido')
df['Idade'] = df['Idade'].fillna(0)


Opção 1: Preencher o valor vazio (.fillna())  

É a melhor opção se você não quiser perder a linha inteira. Podemos substituir o valor vazio pela média de idade dos outros funcionários ou por um valor padrão (como 0 ou -1):

````python
# Criando uma cópia para não alterar o DataFrame original
df_preenchido = df.copy()

# Calcula a média das idades (arredondando para número inteiro)
media_idades = int(df_preenchido['Idade'].mean())

print(f"Média das idades: {media_idades}")

# Preenche o valor vazio da Mariana com a média calculada
df_preenchido['Idade'] = df_preenchido['Idade'].fillna(media_idades)

print("--- Dados após preencher a idade vazia com a média ---")
print(df_preenchido)
````


Opção 2: Remover a linha que contém valores vazios (.dropna())

Útil se o dado ausente for crítico e a linha não puder ser aproveitada sem ele:

````python
# Remove qualquer linha do DataFrame que possua pelo menos um valor vazio
df_sem_nulos = df.dropna()

print("--- Dados após remover linhas com valores vazios ---")
print(df_sem_nulos) # A linha da Mariana sumirá neste resultado
````

### 5. Criação e Modificação de Colunas

Você pode aplicar lógica para criar novas informações a partir das colunas existentes.

```python
# Criando uma nova coluna calculada
df['Idade_em_5_anos'] = df['Idade'] + 5

# Utilizando uma condição para criar uma coluna (Categoria)
# Se idade >= 18, 'Maior de Idade', senão 'Menor de Idade'
import numpy as np
df['Status'] = np.where(df['Idade'] >= 18, 'Maior de Idade', 'Menor de Idade')
```

````python
import pandas as pd

print(f"*"*80)
# 1. Carregamos o arquivo e tratamos os dados nulos (conforme passos anteriores)
df = pd.read_csv('Arquivos_CSV/dados_funcionarios.csv', sep=';', encoding='utf-8')
# carrega o arquivo original 
media_idades = int(df['Idade'].mean())
# calcula a média das idades
df['Idade'] = df['Idade'].fillna(media_idades) 
#substitui os valores nulos pela média
print("--- DataFrame após tratamento de valores nulos ---")
print(df)

print(f"*"*80)
# ----------------------------------------------------
# PASSO 1: Criando a coluna de Bônus (Item 5)
# ----------------------------------------------------
# Criamos uma nova coluna chamada 'Bonus_R$' que recebe 10% (0.10) do 'Salario_R$'
df['Bonus_R$'] = df['Salário_R$'] * 0.10

# Salário total (salário + bônus)
df['Salario_Total_R$'] = df['Salário_R$'] + df['Bonus_R$']

# Criamos também uma coluna com o Salário Total (Salário + Bônus)
df['Salario_Total_R$'] = df['Salário_R$'] + df['Bonus_R$']

print("--- DataFrame com as novas colunas de bônus ---")
print(df)


print(f"*"*80)
# ----------------------------------------------------
# PASSO 2: Salvando o resultado em um novo CSV
# ----------------------------------------------------
# Usamos o método .to_csv() para exportar o arquivo tratado
df.to_csv(
    'Arquivos_CSV/dados_funcionarios_limpos.csv', 
    sep=';',           # Mantém o padrão brasileiro de ponto e vírgula
    index=False,       # IMPORTANTE: False impede que o Pandas crie uma coluna extra com os números das linhas (0, 1, 2...)
    encoding='utf-8'   # Garante a preservação dos acentos no novo arquivo
)

print("\n[Sucesso] O arquivo 'dados_funcionarios_limpos.csv' foi gerado com sucesso!")

print(f"*"*80)
````


💡 Detalhes importantes:  

• index=False: Se você esquecer de colocar isso, toda vez que salvar e reabrir o arquivo, o Pandas criará uma coluna sem nome no início (como Unnamed: 0), sujando o seu arquivo CSV.  
• Operações Diretas: O Pandas permite fazer operações matemáticas direto na coluna (df['Coluna'] * 0.10) de forma extremamente rápida, aplicando o cálculo linha por linha automaticamente (conceito chamado de vetorização).

📝 Enunciados dos Exercícios


Exercício 1 (Básico: Leitura, Tratamento e Nova Coluna)

Escreva um código em Python utilizando Pandas que:
1. Leia o arquivo vendas_loja.csv configurando o separador correto (;).
2. Preencha o valor vazio da coluna Quantidade com o número 1.
3. Crie uma nova coluna chamada Faturamento que seja a multiplicação da Quantidade pelo Preco_Unitario.

Exercício 2 (Intermediário: Filtro e Ordenação)

Utilizando o DataFrame resultante do Exercício 1, faça o seguinte:
1. Filtre os dados para exibir apenas as vendas da região 'Nordeste' OU que tenham o produto 'Notebook'.
2. Ordene esse resultado para mostrar os maiores faturamentos no topo (Faturamento de forma decrescente).

Exercício 3 (Intermediário: Agrupamento e Exportação)

Utilizando o DataFrame resultante do Exercício 1, faça o seguinte:
1. Agrupe os dados por Regiao e descubra a soma total do Faturamento de cada uma delas.
2. Salve o resultado desse agrupamento em um novo arquivo chamado faturamento_por_regiao.csv (lembre-se de configurar o separador correto e remover o índice).


🔑 Gabarito / Resolução Comentada


````python
import pandas as pd

print(f"*"*80)

df = pd.read_csv('vendas_loja.csv', sep=';', encoding='utf-8')

df['Quantidade'] = df['Quantidade'].fillna(1)

df['Faturamento'] = df['Quantidade'] * df['Preco_Unitario']

print("--- DataFrame do Exercicio 1 Concluído")
print(df.head(10))

print(f"*"*80)

````

````python
condicao = (df['Regiao'] == 'Nordeste') | (df['Produto'] == 'Notebook')

df_filtrado = df[condicao]

df_ordenado = df_filtrado.sort_values(by='Faturamento', ascending=False)

print("---Resultado do Exercício 2 (FIltrado e Ordenado)---")
print(df_ordenado)
print("/n" + "="*80 + "/n")
````

````python
print(f"*"*40)
agrupado_regiao = df.groupby('Regiao') ['Faturamento'].sum().reset_index()

print("--- Resultado do Exercicio 3 ---")
print(agrupado_regiao)
print(f"*"*40)
agrupado_regiao.to_csv('faturamento_po_regiao.csv', sep=';', index=False)
````
