#### Pandas em Python


O Pandas é uma biblioteca do Python utilizada para manipular, organizar, limpar e analisar dados. É muito útil para trabalhar com planilhas, arquivos CSV, tabelas de vendas, cadastros de clientes, relatórios financeiros e conjuntos de dados em geral.

Este tutorial foi organizado em etapas, com explicações, exemplos comentados e exercícios práticos, podendo ser utilizado como material de apoio para aulas introdutórias de Python.

#### O que é o Pandas?
O Pandas trabalha principalmente com duas estruturas de dados:

<B>Series</B>

É uma estrutura unidimensional, semelhante a uma única coluna de uma planilha.

Exemplo: uma lista de preços, nomes ou idades.


Lista da feira: Limão, maça, uva,  mandioca

<B>DataFrame</B>

É uma estrutura bidimensional, organizada em linhas e colunas, semelhante a uma planilha do Excel.

Exemplo: uma tabela com produtos, quantidades, preços e categorias.

Exemplo:  

| Nome   | Idade |
| ------ | ----- |
| Ana    | 22    |
| João   | 30    |
| Carlos | 25    |

#### Criando um DataFrame
Podemos criar uma tabela diretamente no Python, sem precisar importar um arquivo externo.

```python
import pandas as pd

dados = {
    "Nome": ["Ana", "João", "Carlos"],
    "Idade": [22, 30, 25]
}

df = pd.DataFrame(dados)

print(df)
```

Entendendo o código:
1. Criamos um dicionário chamado dados.
2. Cada chave do dicionário representa uma coluna da tabela.
3. Os valores de cada chave formam os dados das linhas.
4. pd.DataFrame(dados) transforma o dicionário em uma tabela.
5. print(df) exibe a tabela no console.
O nome df é uma convenção comum para representar um DataFrame.

#### Visualizando os dados  
O Pandas possui comandos para visualizar rapidamente o conteúdo de uma tabela.

```python

print(f"*"*80)
print(f"**Informações sobre o DataFrame:**")
print(f"Exibe as cinco primeiras linhas")
print(dfbase.head())


print(f"*"*80)
print(f"Exibe as duas primeiras linhas")
print(dfbase.head(2))


print(f"*"*80)
print(f"Exibe as duas últimas linhas")
print(dfbase.tail(2))


print(f"*"*80)
print(f"Exibe os nomes das colunas")
print(dfbase.columns)


print(f"*"*80)
print(f"Exibe os tipos de dados e valores não nulos")
print(dfbase.info())


print(f"*"*80)
print(f"Apresenta estatísticas das colunas numéricas")
print(dfbase.describe())

print(f"*"*80)
```

Comando	Finalidade  
- head() -->	Mostra as primeiras linhas  
- tail() -->	Mostra as últimas linhas  
- shape -->	Retorna quantidade de linhas e colunas
- columns ->>	Mostra os nomes das colunas
- info() ->>	Apresenta informações sobre a estrutura
- describe() -->	Calcula estatísticas descritivas

Por exemplo, df.shape retorna (5, 4), pois nossa tabela contém cinco linhas e quatro colunas.

#### Selecionando colunas e linhas  
É possível selecionar apenas as informações necessárias para uma análise.

```python
print(f"*"*80)
print(f" Selecionar somente a coluna Produto")
print(dfbase["Produto"])


print(f"*"*80)
print(f" Selecionar somente a coluna Preço")
print(dfbase["Preço"])

print(f"*"*80)
print(f" Selecionar duas colunas")
print(dfbase[["Produto", "Preço"]])

print(f"*"*80)
print(f" Selecionar a primeira linha pela posição")
print(dfbase.iloc[0])

print(f"*"*80)
print(f" Selecionar as três primeiras linhas")
print(dfbase.iloc[0:3])
```

**Observe a diferença:** 
- df["Produto"] --> retorna uma única coluna, no formato Series.
- df[["Produto", "Preço"]] --> retorna um DataFrame com duas colunas.
- df.iloc[0] --> retorna a primeira linha, pois as posições começam em zero.
- df.iloc[0:3] --> retorna as linhas das posições zero, um e dois.

#### Filtrando dados com condições  
Uma das funcionalidades mais úteis do Pandas é filtrar informações com base em condições.

Imagine que queremos descobrir quais produtos custam mais de R$ 500.

```python
print(f"*"*80)
print(f" Filtrar produtos com preço superior a 500")
# Filtra produtos com preço superior a 500
filtro = dfbase[dfbase["Preço"] > 500]

print(filtro)
print(f"*"*80)
```

**Sua Vez Mostre:**
- Produtos com quantidade menor que 10
- Produtos da categoria Informática
- Produtos com preço entre 100 e 800, inclusive


```python
print(f"*"*80)
print(f"Filtrar produtos informáticos com preço entre 100 e 800, com quantidade menor que 10")

filtro = dfbase[dfbase["Quantidade"] < 10]

print(filtro)
print(f"*"*80)
```

Atenção aos operadores:
- ( ==) verifica igualdade.
- ( > ) verifica se um valor é maior.
- ( < ) verifica se um valor é menor.
- ( & ) combina condições que devem ser verdadeiras simultaneamente.
- ( | ) combina condições em que pelo menos uma deve ser verdadeira.

Ao combinar condições com & ou |, utilize parênteses em cada comparação.

#### Criando e alterando colunas  
Podemos calcular informações novas a partir dos dados existentes.

Por exemplo, vamos calcular o valor total de estoque de cada produto.

```python

print(f"*"*80)
print(f" Valor total de cada produto em estoque")
# Calcula o valor total de cada produto em estoque
dfbase["Valor Estoque"] = dfbase["Quantidade"] * dfbase["Preço"]

print(dfbase)
print(f"*"*80)
```

A nova coluna será calculada multiplicando a quantidade pelo preço unitário.

Exemplo: o teclado tem 10 unidades e custa R$ 120,00.

\(10 \times 120 = 1.200\)

Portanto, o valor total do estoque de teclados será R$ 1.200,00.

Também podemos alterar valores de uma coluna:

```python
print(f"*"*80)
print(f" Aumenta em 10% todos os preços")
# Aumenta em 10% todos os preços
dfbase["Preço"] = dfbase["Preço"] * 1.10

print(dfbase[["Produto", "Preço"]])
print(f"*"*80)
```

Esse comando altera os preços originais em memória. Em um projeto real, é importante verificar se a alteração deve ser permanente e salvar os resultados em um arquivo quando necessário.

#### Ordenando os dados
Para classificar os produtos por preço, quantidade ou outra coluna, utilizamos sort_values().

```python
print(f"*"*80)
print(f" Ordena do menor para o maior preço")
# Ordena do menor para o maior preço
print(dfbase.sort_values(by="Preço"))



print(f"*"*80)
# Ordena do maior para o menor preço
print(dfbase.sort_values(by="Preço", ascending=False))


print(f"*"*80)
# Ordena pela quantidade, em ordem crescente
print(dfbase.sort_values(by="Quantidade"))

print(f"*"*80)
```

O parâmetro ascending=False inverte a ordem da classificação.

#### Realizando cálculos e análises   
O Pandas facilita a obtenção de indicadores numéricos.

```python
print(f"*"*80)
print(f" Soma das quantidades")
print(dfbase["Quantidade"].sum())

print(f"*"*80)
print(f" Média dos preços")
print(dfbase["Preço"].mean())

print(f"*"*80)
print(f" Maior preço")
print(dfbase["Preço"].max())

print(f"*"*80)
print(f" Menor preço")
print(dfbase["Preço"].min())

print(f"*"*80)
print(f" Quantidade de produtos cadastrados")
print(dfbase["Produto"].count())

print(f"*"*80)
# Valor total do estoque
dfbase["Valor Estoque"] = dfbase["Quantidade"] * dfbase["Preço"]

print(f" Valor total do estoque: R$ {dfbase['Valor Estoque'].sum():,.2f}")
print(f"\n{'*'*80}")
```

**Esses comandos permitem responder perguntas como:**
- Qual é o valor médio dos produtos?
- Qual produto possui o maior preço?
- Quantas unidades existem no estoque?
- Qual é o valor financeiro total do estoque?

#### Agrupando dados com groupby()  
O groupby() permite agrupar registros por uma categoria e calcular indicadores para cada grupo.

Por exemplo, queremos saber a quantidade total de produtos em cada categoria.

```python
print(f"*"*80)
print(f" Soma a quantidade de produtos por categoria")

resultado = dfbase.groupby("Categoria")["Quantidade"].sum()

print(resultado)
print(f"*"*80)
```

Também podemos calcular várias métricas ao mesmo tempo:

```python
print(f"*"*80)
# Cria a coluna de valor total
dfbase["Valor Estoque"] = dfbase["Quantidade"] * dfbase["Preço"]

# Agrupa por categoria e calcula indicadores
resumo = dfbase.groupby("Categoria").agg( # aqui estamos agrupando por categoria e calculando indicadores
    Total_Unidades=("Quantidade", "sum"), # este é o total de unidades por categoria
    Preco_Medio=("Preço", "mean"), # este é o preço médio por categoria
    Valor_Total=("Valor Estoque", "sum") # este é o valor total do estoque por categoria
)

print(resumo)
```

Esse recurso é bastante utilizado na elaboração de relatórios gerenciais, análises de vendas e acompanhamento de estoque.

Exemplo completo: análise de estoque  
Agora vamos reunir os principais comandos em um programa.

```python

import pandas as pd

# 1. Cria os dados
dados = {
    "Produto": ["Teclado", "Mouse", "Monitor", "Cadeira", "Mesa"],
    "Categoria": [
        "Informática",
        "Informática",
        "Informática",
        "Escritório",
        "Escritório"
    ],
    "Quantidade": [10, 15, 5, 8, 3],
    "Preço": [120.00, 60.00, 950.00, 700.00, 850.00]
}

# 2. Cria o DataFrame
df = pd.DataFrame(dados)

# 3. Calcula o valor do estoque
df["Valor Estoque"] = df["Quantidade"] * df["Preço"]

# 4. Exibe todos os produtos
print(f"*"*80)
print("\n--- ESTOQUE COMPLETO ---")
print(df)

# 5. Exibe produtos com quantidade inferior a 6
print(f"*"*80)
print("\n--- ESTOQUE BAIXO ---")
print(df[df["Quantidade"] < 6])

# 6. Exibe os produtos mais caros primeiro
print(f"*"*80)
print("\n--- PRODUTOS POR PREÇO ---")
print(df.sort_values(by="Preço", ascending=False))

# 7. Calcula o valor total do estoque
print(f"*"*80)
total = df["Valor Estoque"].sum()

print("\nValor total do estoque: R$", round(total, 2))

# 8. Gera um resumo por categoria
print(f"*"*80)
print("\n--- RESUMO POR CATEGORIA ---")
print(df.groupby("Categoria")["Valor Estoque"].sum())

print(f"*"*80)
```
